# 📘 Análisis de Producto

Este notebook presenta rankings de equipos, tallas y modelos para guiar decisiones de compra.

```python
import pandas as pd
import matplotlib.pyplot as plt

csv_path = "data_jerseys_cleaned.csv"

df = pd.read_csv(csv_path, parse_dates=['Fecha de compra','Fecha llegada','Fecha de venta'])
df = df.drop_duplicates()

# Top equipos por ventas
top_equipos = df['equipo'].value_counts().head(10)
# tallas más vendidas
top_tallas = df['Talla'].value_counts().head(10)
# modelos más vendidos (usar columna Jugador o version)
top_modelos = df['Jugador'].value_counts().head(10)

# modelos más lentos: calcular mediana de dias entre llegada y venta por modelo
mask = df['Fecha de venta'].notna() & df['Fecha llegada'].notna()
df_time = df[mask].copy()
df_time['dias_venta'] = (df_time['Fecha de venta'] - df_time['Fecha llegada']).dt.days
modelos_lentos = df_time.groupby('Jugador')['dias_venta'].median().sort_values(ascending=False).head(10)

# rentabilidad por modelo
rent_modelo = df.groupby('Jugador').agg(ventas=('Jugador','count'), ganancia_total=('Ganancia','sum'))
rent_modelo['ganancia_media'] = rent_modelo['ganancia_total'] / rent_modelo['ventas']

# mostrar resultados
top_equipos, top_tallas, top_modelos, modelos_lentos.head(), rent_modelo.sort_values('ganancia_media',ascending=False).head(10)
```
