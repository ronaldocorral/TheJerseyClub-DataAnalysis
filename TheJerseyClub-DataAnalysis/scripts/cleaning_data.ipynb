```python
import pandas as pd
from google.colab import files

data_jerseys=pd.read_csv('/raw_datajerseys.csv', encoding='latin1')


print(data_jerseys.head(10))
```

```python
# Convert 'Fecha llegada' to datetime, coercing errors to NaT
data_jerseys['Fecha llegada'] = pd.to_datetime(data_jerseys['Fecha llegada'], errors='coerce')

# Rellenar NaN en 'Fecha llegada' según tabla Pedido -> Fecha de llegada
fechas_por_pedido = {
    7:  '9/14/2025',
    8:  '10/24/2025',
    10: '11/1/2025',
    11: '11/17/2025',
    12: '12/14/2025',
    13: '1/6/2026',
    14: '1/23/2026',
    15: '2/15/2026',
    16: '2/5/2026',
    17: '2/12/2026',
    18: '2/18/2026',
}

# Convertimos el diccionario a datetime para que asigne fechas reales
fechas_por_pedido = {k: pd.to_datetime(v, errors='coerce') for k, v in fechas_por_pedido.items()}

# Solo rellenar donde esté nulo, usando el Pedido como llave
mask = data_jerseys['Fecha llegada'].isna()
data_jerseys.loc[mask, 'Fecha llegada'] = data_jerseys.loc[mask, 'Pedido'].map(fechas_por_pedido)

# Replace 'Pedido' number 19 with 18
data_jerseys['Pedido'] = data_jerseys['Pedido'].replace(19, 18)

# Add 'Proveedor' column
data_jerseys['Proveedor'] = 'Alibaba'
data_jerseys.loc[data_jerseys['Pedido'].isin([16, 17, 18]), 'Proveedor'] = 'Tiktok'


# Display the modified DataFrame to verify the changes
print(data_jerseys)
```

```python


# Asegurarse de que 'Fecha de compra' y 'Fecha llegada' sean tipo datetime
data_jerseys['Fecha de compra'] = pd.to_datetime(data_jerseys['Fecha de compra'], errors='coerce')
data_jerseys['Fecha llegada'] = pd.to_datetime(data_jerseys['Fecha llegada'], errors='coerce')

# Crear/recalcular la columna 'Tiempo de entrega' en días
data_jerseys['Tiempo en llegar'] = (data_jerseys['Fecha llegada'] - data_jerseys['Fecha de compra']).dt.days.fillna(0).astype(int)

# Reemplazar valores nulos en 'Jugador' por 'Sin nombre'
data_jerseys['Jugador'] = data_jerseys['Jugador'].fillna('Sin nombre')

# Limpiar los nombres de las columnas para eliminar espacios en blanco
data_jerseys.columns = data_jerseys.columns.str.strip()

# Definir el orden deseado de las columnas
desired_column_order = [
    'Pedido',
    'Fecha de compra',
    'Fecha llegada',
    'Tiempo en llegar',
    'Fecha de venta',
    'equipo',
    'deporte',
    'version',
    'temporada',
    'Jugador',
    'Talla',
    'Proveedor',
    'Costo',
    'Costo de envio',
    'Costo total',
    'Venta',
    'Ganancia'
]

# Reordenar las columnas del DataFrame. Esto también eliminará la columna 'Modelo' si no está en la lista.
# Primero, aseguramos que todas las columnas en `desired_column_order` existan en el DataFrame.
# Si una columna no existe, se generará un error, lo cual es deseable para identificar si falta algo.
# Si 'Modelo' estuviera todavía presente, no se incluiría en el nuevo DataFrame al reordenar.
data_jerseys = data_jerseys[desired_column_order]

# Mostrar las primeras filas del DataFrame con el nuevo orden de columnas
display(data_jerseys.head())

print(data_jerseys.info)
```

```python
# Convert 'Fecha de venta' to datetime
data_jerseys['Fecha de venta'] = pd.to_datetime(data_jerseys['Fecha de venta'], errors='coerce')

# List of columns to convert to numeric
numeric_cols = ['Costo', 'Costo de envio', 'Costo total', 'Venta', 'Ganancia']

# Iterate through the columns and convert them to numeric
for col in numeric_cols:
    # Remove '$' and any spaces, then convert to numeric, coercing errors to NaN
    data_jerseys[col] = data_jerseys[col].astype(str).str.replace('$', '', regex=False).str.strip()
    data_jerseys[col] = pd.to_numeric(data_jerseys[col], errors='coerce')

# Display data_jerseys.info() to verify the new data types
data_jerseys.info()
```
